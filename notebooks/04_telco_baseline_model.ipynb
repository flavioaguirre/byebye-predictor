### Importing libraries and setup

```python
from notebooks_setup import PROJECT_ROOT

from src.utils import get_logger
from src.data_loader import load_csv, preview_df
from src.eda import dataframe_overview
from src.model_builder import ModelBuilder
from src.model_evaluation import ModelEvaluatorConfig, ModelEvaluator 

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

<br>

<hr>

### Loading the dataset and setup

```python
logger = get_logger(__name__)
```

```python
config = ModelEvaluatorConfig()

logger.info("Configuration applied:")
logger.info(f"Output directory: {config.out_dir}")
logger.info(f"Plotting style: {config.style}")
```

```python
path_df_processed  = "data/processed/telco_churn_curated.csv"
df_processed = load_csv(path_df_processed)

logger.info(f"Shape: {df_processed.shape}")
preview_df(df_processed)
```

<br>

### Target distribution

We remember that in our [Reports]("../reports/figures/eda/target_distribution.png") we already have our target variable distributed.
<div style="text-align:center">
    <img src="../reports/figures/eda/target_distribution.png" width="600" alt="target distribution">
</div>

We looked at the ratio of customers who churned versus those who stayed.
This is important because imbalance can affect model performance.

```python
# Target load
path_target_processed = "data/processed/target_processed.csv"
target_processed = load_csv(path_target_processed)
```

<br>

### Definition of X and Y

```python
X = df_processed
y = target_processed

logger.info("Set X and y defined correctly!")
```

```python
RANDOM_STATE = 42

logger.info("RANDOM_STATE set correctly")
```

<br>

### Train/test split

```python
logger.info("Separating data into training and test subsets...")
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, 
    stratify=y, 
    random_state=RANDOM_STATE
)

logger.info("Data correctly separated into training and test subsets!")
logger.info(f"Train size: {X_train.shape}, Test size: {X_test.shape}")
```

<br>

### Construction of the models

We use our aforementioned pipeline (ModelBuilder) to build our models and evaluate them to obtain the one with the best statistics.

```python
builder = ModelBuilder(task='classification', random_state=RANDOM_STATE)
```

The method uses cross-validation internally on X_train, y_train.

```python
models_evaluation = builder.evaluate_models(X_train, y_train, cv=7)
```

```python
evaluation_of_results_df = pd.DataFrame(models_evaluation)

logger.info("Results of all models applying the cross-validation method (7-fold average):")
evaluation_of_results_df
```

<br>

### Construction of the baseline

#### Selecting Best Model

```python
# We select the best model based on F1-score
builder.select_best_model(metric='test_f1')
best_model_name = builder.best_model_name_
logger.info(f"The best model according to F1-score is: '{best_model_name}'")
```

#### Train that specific model on the full training set

```python
final_model = builder.train_final_model(best_model_name, X_train, y_train)

logger.info(f"Final model '{best_model_name}' trained with {X_train.shape[0]} samples.")
```

<br>

### Final evaluation with the winning model on the test set

Now we use our dedicated module (ModelEvaluator) to analyze this model.

```python
evaluator = ModelEvaluator(task="classification", save_dir=config.out_dir)
```

```python
y_pred = final_model.predict(X_test)
```

#### Metrics

**We obtain that:**

```python
metrics = evaluator.evaluate(final_model, X_test, y_test, run_name="baseline_no_features")

evaluator.compare_runs().T
```

#### Classification report

```python
logger.info(f"Final classification report for '{best_model_name}':")
print(classification_report(y_test, y_pred))
```

This report indicates that the model's performance is balanced, although with marked differences between the prediction of customers who do not churn (Class 0) and those who do churn (Class 1):

**Class 0 (No Churn):**

* Accuracy (0.90) → When the model predicts that a customer will not churn, it is correct in 90% of cases.

* Recall (0.72) → It correctly identifies 72% of customers who do not churn.

* F1-score (0.80) → Good balance between precision and recall.

**Class 1 (Churn):**

* Accuracy (0.51) → Of all customers predicted as churn, only 51% actually churn (high level of false positives).

* Recall (0.78) → The model detects 78% of customers who actually churn, which is positive in terms of churn reduction.

* F1-score (0.61) → Moderate performance, limited by low precision.

**Overall metrics:**

* Accuracy (0.74) → 74% of predictions are correct.

* Macro Avg (0.71) → The average performance between both classes is balanced, although lower in the churn class.

* Weighted Avg (0.75) → Reflects the actual weight of each class in the dataset, confirming that the model is stronger at predicting "No Churn."

#### Confusion Matrix

```python
confusion_matrix = evaluator.plot_confusion_matrix(
    y_test, 
    y_pred, 
    labels=["No Churn", "Churn"],
    title=f"Best baseline confusion matrix ({best_model_name})",
    filename=f"confusion_matrix_best_baseline_{best_model_name}.png"
)
```

#### ROC Curve

```python
roc_curve =evaluator.plot_roc_curve(
    final_model, 
    X_test, 
    y_test,
    title=f"Best baseline ROC Curve ({best_model_name})",
    filename=f"roc_curve_best_baseline_{best_model_name}.png"
)
```

#### Precision-Recall Curve

```python
precision_recall = evaluator.plot_precision_recall(
    final_model, 
    X_test, 
    y_test,
    title=f"Best baseline Precision-Recall Curve ({best_model_name})",
    filename=f"precision_recall_curve_best_baseline_{best_model_name}.png"
)
```

### Final results of this laboratory

##### Classification Report

We can deduce that the ``logistic regression`` baseline achieves solid initial performance (AUC 0.84 and accuracy 74%), with clear strength in identifying customers who don't churn and high recall for those who do. However, the low churn accuracy implies false positives: the model overestimates the number of customers who would churn.

##### Confusion Matrix

<div style="text-align:center">
    <img src="../reports/figures/baseline-model-telco/confusion_matrix_best_baseline_logistic_regression.png" width="600" alt="Confusion Matrix baseline">
</div>

The model is quite good at detecting customers who are staying (high TN = ["No churn","No churn"]).

It has a relatively low FN (["No churn","Churn"]) (81), which is positive because in churn, the most expensive thing is not detecting a customer who actually leaves.

However, there is a significant number of FPs (False Positive ["churn","no churn"] = 287) → the model "over-triggers alarms" by predicting that some customers will leave when in reality they won't.

This means that if a company decides to invest in customer retention based on this model, it will waste efforts on customers who weren't going to leave.

##### ROC Curve

<div style="text-align:center">
    <img src="../reports/figures/baseline-model-telco/roc_curve_best_baseline_logistic_regression.png" width="600" alt="Confusion Matrix baseline">
</div>

With an AUC of 0.84, the model has an excellent ability to distinguish between churning and retaining customers.

This means that, given one customer who churns and one who doesn't, there is an 84% probability that the model will assign a higher probability of churn to the churner.

In the business world, this indicates that logistic regression is already a solid foundation, although there is still room for improvement with feature engineering or more complex models.

##### Precision-Recall Curve

<div style="text-align:center">
    <img src="../reports/figures/baseline-model-telco/precision_recall_curve_best_baseline_logistic_regression.png" width="600" alt="Confusion Matrix baseline">
</div>

In the graph, the initial accuracy is quite good (~0.75-0.8 on average), but it begins to drop sharply after ~0.5 recall.
This implies that the model cannot reliably detect all churners, but it can be useful in scenarios where you want to prioritize customers with a high probability of churning.
